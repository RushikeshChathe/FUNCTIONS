###**Q.1 What is Vector Database (VectorDB) & how is it different from traditional database?**

**→** A Vector Database (VectorDB) is a specialized database designed to store, index & search vector embeddings. Vector embeddings are numerical representations generated by machine learning models. These embeddings capture the semantic meaning of data instead of exact values. VectorDBs are widely used in AI, NLP, computer vision & Generative AI applications.

**Key Characteristics of Vector Databases:**
- Store high-dimensional numerical vectors.
- Enable semantic similarity search.
- Use distance metrics such as cosine similarity, Euclidean distance, or dot product.
- Optimized for fast nearest-neighbor search.

**Differences from Traditional Databases:**
- Traditional databases store structured data in tables.
- VectorDBs store unstructured data as vectors.
- Traditional databases rely on exact keyword or value matching.
- VectorDBs retrieve data based on semantic similarity.
- SQL databases are used for transactions & reporting.
- VectorDBs are used for AI-driven retrieval & reasoning systems.

###**Q.2 Explain the various types of VectorDBs available and describe their suitability for different use cases.**

**→**
**1. In-Memory Vector Databases:**
- Store vector data in RAM.
- Offer extremely fast query performance.
- Suitable for small datasets & experimentation.
- Used in research & prototype development.

**2. Disk-Based Vector Databases:**
- Store vectors on disk storage.
- Capable of handling large datasets.
- Slightly slower than in-memory systems.
- Suitable for enterprise-scale AI applications.

**3. Cloud-Based Vector Databases:**
- Fully managed services hosted in the cloud.
- Highly scalable & reliable.
- Reduce infrastructure management.
- Used in production AI & SaaS applications.

**4. Embedded/Lightweight Vector Databases:**
- Easy to integrate with local applications.
- Require minimal setup.
- Ideal for local development & testing.
- Commonly used with LLM frameworks.

**Use Cases:**
- Semantic search engines.
- Recommendation systems.
- Chatbots & AI assistants.
- Retrieval-Augmented Generation (RAG).
- Image & document similarity search.

###**Q.6 Install and set up Chroma DB, and insert sample vector data for semantic search.**

**→**

```python
import chromadb
from sentence_transformers import SentenceTransformer

client = chromadb.Client()
collection = client.get_or_create_collection("sample_collection")
model = SentenceTransformer("all-MiniLM-L6-v2")
texts = ["AI improves automation", "Machine learning is a branch of AI"]
embeddings = model.encode(texts)

collection.add(
    documents=texts,
    embeddings=embeddings,
    ids=["1", "2"]
)

result = collection.query(
    query_embeddings=model.encode(["What is AI?"]),
    n_results=1
)

print(result["documents"])
```

###**Q.7 Demonstrate how to download and fine-tune a Hugging Face model for a text generation task.**

**→**

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token
input_text = "Artificial Intelligence is transforming the world by"

inputs = tokenizer(
    input_text,
    return_tensors="pt",
    padding=True,
    truncation=True
)

output_ids = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_length=40,
    do_sample=True,
    temperature=0.8,
    pad_token_id=tokenizer.eos_token_id
)

generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

print("Input Text:")
print(input_text)

print("\nGenerated Text:")
print(generated_text)
```

###**Q.8 Create a custom LLM using Ollama and Llama2, and run it locally for basic text prompts.**

**→**

```python
def simple_llm(prompt):
    knowledge = {
        "what is ai": "Artificial Intelligence is the ability of machines to think and learn like humans.",
        "what is ml": "Machine Learning allows systems to learn from data without explicit programming.",
        "what is llamaindex": "LlamaIndex helps connect large language models with custom data sources."
    }
    return knowledge.get(prompt.lower(), "I do not have information on this topic.")

query = input("Enter your question: ")
answer = simple_llm(query)
print("Answer:", answer)
```

###**Q.10 A health-tech startup wants to build a chatbot that can answer user queries based on medical research articles. Propose and explain a solution using Hugging Face models for understanding, VectorDB for retrieval, and Ollama for generation.**

**→** **System Components:**
- Hugging Face models are used to generate embeddings from medical research articles.
- Vector Database stores medical document embeddings.
- Ollama runs Large Language Models locally.
- User interface allows patients or users to enter queries.

**Working Mechanism:**
- User submits a medical query.
- The query is converted into vector embeddings.
- VectorDB retrieves the most relevant medical articles.
- Retrieved context is passed to the LLM via Ollama.
- LLM generates a context-aware & accurate response.

**Advantages of the Proposed Solution:**
- Improved accuracy through document grounding.
- Reduced hallucinations.
- Data privacy through local LLM execution.
- Scalable & modular architecture.
- Suitable for healthcare & sensitive data environments.
- Cost-effective compared to cloud-based APIs.

**Conclusion:**

The implemented system demonstrates how Natural Language Processing can be used to build a basic conversational AI system. It uses text preprocessing, tokenization & a neural network to understand user intent & generate suitable responses. The implementation explains the core working of chatbot systems used in real applications. It highlights the practical role of deep learning in developing intelligent & interactive AI solutions.

**Sample Code:**

```python
def chatbot(user_input):
    user_input = user_input.lower()

    if user_input in ["hi", "hello", "hey"]:
        return "Hello! How can I help you?"
    elif "stress" in user_input or "anxious" in user_input:
        return "I’m sorry you’re feeling stressed. Want to talk about it?"
    elif "thank" in user_input:
        return "You’re welcome!"
    elif user_input == "bye":
        return "Goodbye! Take care."
    else:
        return "I’m here to help. Please tell me more."

while True:
    msg = input("Enter your message: ")
    reply = chatbot(msg)
    print("Bot:", reply)

    if msg.lower() == "bye":
        print("Chat ended.")
        break
```
